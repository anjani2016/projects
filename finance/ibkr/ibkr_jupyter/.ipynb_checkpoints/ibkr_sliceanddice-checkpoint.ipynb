```python
df_cour_contract_details['validExchanges_list'] = ( 
    df_cour_contract_details['validExchanges'] 
        .str.split(',') 
        .apply(lambda lst: [x.strip() for x in lst]) 
)
```

```python
df_cour_contract_details['validExchanges_list']
```

# 1. Explode validExchanges_list into multiple rows

```python
df_exploded = df_cour_contract_details.explode('validExchanges_list')
```

```python
df_exploded
```

```python
df_exploded['validExchanges_list'] = df_exploded['validExchanges_list'].str.strip()
```

# 2. Pivot validExchanges_list into multiple columns

```python
df_split = df_cour_contract_details['validExchanges_list'].apply(pd.Series)
```

```python
df_split 
```

```python
df_split1 = df_split.rename(columns=lambda i: f'exch_{i+1}')
```

```python
df_split1
```

```python
df_with_columns = df_cour_contract_details.join(df_split1)
```
